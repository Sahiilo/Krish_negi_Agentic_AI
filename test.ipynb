```python
from typing_extensions import TypedDict,Annotated
import operator

from langchain_core.messages import AnyMessage, HumanMessage, AIMessage
```

```python
class GraphState(TypedDict):
    messages: Annotated[list[AnyMessage], operator.add] 
```

```python
def llm_call(state: GraphState) -> dict: # This function receive an state object 
    """Call the LLM using conversation messages and append AI response."""
    response = chat_llm.invoke(state["messages"]) # AI message
    return{
        "messages": [response] # This return the updated state 
    }
```

Each node return partial state, not the full state
The graph merges this into the main state dictionary.

```python
def token_counter(state: GraphState)-> dict:
    """Count token (simple word count) in th AI message"""
    last_msg = state["messages"][-1] # This give the most recent message(usually LLM reply)
    text = last_msg.content  # This take the row text of the messages. 
    token_number = len(text.split()) # split into the spaces. 
    summary = f"Total token number in the generated answer is {token_number}"
    return{
        "messages": [AIMessage(content=summary)] # Create a new AI message containing the summary. 
    }
```
